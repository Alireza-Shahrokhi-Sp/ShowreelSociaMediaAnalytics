```python
import vertexai
from vertexai.generative_models import GenerativeModel

def analyze_comment_sentiment(comment_text: str):
    # 1. Initialize the Vertex AI SDK with your GCP project details
    # Replace 'your-gcp-project-id' with your actua
    #l Google Cloud Project ID
    vertexai.init(project="gen-lang-client-0792749758", location="us-central1")

    # 2. Load the model (gemini-2.5-flash is fast and cost-effective for text classification)
    model = GenerativeModel("gemini-2.5-flash")

    # 3. Structure your prompt to get a clean, structured response
    prompt = f"""
    Analyze the sentiment of the following social media comment. 
    Respond with exactly one word: Positive, Negative, or Neutral.

    Comment: "{comment_text}"
    """

    # 4. Make the online API call
    response = model.generate_content(prompt)
    
    return response.text.strip()

# --- Example Usage ---
if __name__ == "__main__":
    sample_comment = "I absolutely love how efficient this tool is! Saved me hours of manual tracking."
    
    print("Sending comment to Vertex AI for analysis...")
    result = analyze_comment_sentiment(sample_comment)
    
    print(f"\nComment: {sample_comment}")
    print(f"Sentiment Analysis Result: {result}")
```
